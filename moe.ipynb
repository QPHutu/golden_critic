```python
import torch

def compute_binary_kl(_p, _q):
    _eps = 1e-6
    p = torch.clamp(_p, min=_eps, max=1.0 - _eps)
    q = torch.clamp(_q, min=_eps, max=1.0 - _eps)
    print(p.dtype, q.dtype, p, q, (1 - p) / (1 - q), torch.log((1 - p) / (1 - q)), "|", p / q, torch.log(p / q))
    print(p * torch.log(p / q), (1 - p) * torch.log((1 - p) / (1 - q)))
    return p * torch.log(p / q) + (1 - p) * torch.log((1 - p) / (1 - q))

pp = torch.Tensor([1.0])
qq = torch.Tensor([0.0])
print(compute_binary_kl(pp, qq))
print(compute_binary_kl(qq, pp))
```

```python
import pandas as pd

def make_prompt_critic(row):
    content = row["prompt"][0]["content"]
    answer = row["reward_model"]["ground_truth"]
    critic_content = f"{content}\nThe reference answer is {answer}. Please provide the detailed reasoning steps that lead to this answer.\n"
    return [{"role": "user", "content": critic_content}]


# out_path = '/home/aiops/qiph/verl/data/sanity_test/aime_2025.parquet'
out_path = '/home/aiops/qiph/verl/data/dapo/dapo-math-17k.parquet'
df = pd.read_parquet(out_path)

df["prompt_critic"] = df.apply(make_prompt_critic, axis=1)

df.to_parquet(out_path)
print(f"Saved {len(df)} rows to {out_path}")
df[["prompt", "prompt_critic"]].head(2)
```
