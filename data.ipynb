```python
import json
import pandas as pd
import os
```

```python
def make_data(example, data_source):
    question = example.pop('problem').rstrip()
    answer = example.pop('answer')
    solution = example.pop('solution')

    instruction = "Please reason step by step, and put your final answer within \\boxed{}.\n"
    prompt = f"{question}\n\n{instruction}"
    solution_prompt = f"The reference solution is: {solution}"
    prompt_with_solution = f"{question}\n\n{solution_prompt}\n\n{instruction}"
    answer_prompt = f"The reference answer is: {answer}."
    prompt_with_answer = f"{question}\n\n{answer_prompt}\n\n{instruction}"
    prompt_with_both = f"{question}\n\n{solution_prompt}\\{answer_prompt}\n\n{instruction}"

    example_data = {
        "data_source": data_source,
        "prompt": [{
            "role": "user",
            "content": prompt
        }],
        "prompt_with_solution": [{
            "role": "user",
            "content": prompt_with_solution
        }],
        "prompt_with_answer": [{
            "role": "user",
            "content": prompt_with_answer
        }],
        "prompt_with_both": [{
            "role": "user",
            "content": prompt_with_both
        }],
        "ability": "math",
        "reward_model": {
            "style": "rule",
            "ground_truth": str(answer)
        },
    }
    return example_data


json_path = "/home/aiops/qiph/verl/data/deepscaler/deepscaler.json"

with open(json_path, "r") as f:
    data = json.load(f)

data_list = []
for example in data:
    data_list.append(make_data(example, "sanity_test_deepscaler"))

train_data = pd.DataFrame(data_list)
par_path = "/home/aiops/qiph/verl/data/deepscaler/deepscaler_v1.parquet"
train_data.to_parquet(par_path)
```
