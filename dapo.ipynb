```python
def make_prompt_critic(row):
    messages = row["prompt"]
    answer = row["reward_model"]["ground_truth"]

    prompt = messages[-1]["content"]
    last = prompt.split("\n\n")[-1]
    last_len = len(last)
    prompt_critic = prompt[:-last_len] + f"The reference answer is: {answer}.\n\n" + last

    new_messages = [dict(m) for m in messages]
    new_messages[-1] = {**new_messages[-1], "content": prompt_critic}
    return new_messages


df["prompt_critic"] = df.apply(make_prompt_critic, axis=1)

print(df["prompt_critic"].iloc[0][-1]["content"])
# df.head()

df.to_parquet('/home/aiops/qiph/verl/data/dapo/dapo-math-17k-unique.parquet')
```
